CAPTCHA GENERATION

```python
import random
import string
from PIL import Image, ImageDraw, ImageFont, ImageFilter
from IPython.display import display # Import display for showing images in Colab

def generate_captcha_text(length=8, chars=None):
    """Generate a random CAPTCHA string."""
    if chars is None:
        # Mix of digits, lowercase, uppercase — avoid visually ambiguous chars
        chars = string.digits + string.ascii_lowercase + string.ascii_uppercase
        chars = chars.replace('0', '').replace('O', '').replace('l', '').replace('I', '1')
    return ''.join(random.choices(chars, k=length))


def add_noise_dots(draw, width, height, count=80, color=(0, 0, 255)):
    """Add random scattered dots (like the blue dots in the sample)."""
    for _ in range(count):
        x = random.randint(0, width)
        y = random.randint(0, height)
        radius = random.randint(1, 3)
        draw.ellipse([x - radius, y - radius, x + radius, y + radius], fill=color)


def add_noise_squares(draw, width, height, count=10, color=(0, 0, 0)):
    """Add random small black squares as extra noise."""
    for _ in range(count):
        x = random.randint(0, width - 6)
        y = random.randint(0, height - 6)
        size = random.randint(3, 7)
        draw.rectangle([x, y, x + size, y + size], fill=color)


def generate_captcha(
    text=None,
    width=300,
    height=80,
    bg_color=(255, 255, 255),
    text_color=(0, 0, 0),
    font_size=52,
    font_path=None,          # Path to a .ttf font file; None = PIL default
    blue_dot_count=80,
    black_square_count=12,
    distort=True,
    output_path="captcha.png",
):
    """
    Generate a CAPTCHA image.

    Args:
        text            : The CAPTCHA string (auto-generated if None).
        width / height  : Image dimensions in pixels.
        bg_color        : Background colour (RGB tuple).
        text_color      : Main text colour (RGB tuple).
        font_size       : Point size of the text.
        font_path       : Path to a TrueType/OpenType font file.
        blue_dot_count  : Number of blue noise dots.
        black_square_count: Number of black noise squares.
        distort         : Apply slight per-character vertical jitter.
        output_path     : Where to save the final PNG.

    Returns:
        (image, text) — PIL Image object and the CAPTCHA string used.
    """
    if text is None:
        text = generate_captcha_text()

    img = Image.new("RGB", (width, height), color=bg_color)
    draw = ImageDraw.Draw(img)

    # --- Load font -------------------------------------------------------
    try:
        if font_path:
            font = ImageFont.truetype(font_path, font_size)
        else:
            # Try common system mono fonts before falling back to PIL default
            for candidate in [
                "/usr/share/fonts/truetype/dejavu/DejaVuSansMono-Bold.ttf",
                "/usr/share/fonts/truetype/liberation/LiberationMono-Bold.ttf",
                "/Library/Fonts/Courier New Bold.ttf",
                "C:/Windows/Fonts/courbd.ttf",
            ]:
                try:
                    font = ImageFont.truetype(candidate, font_size)
                    break
                except IOError:
                    continue
            else:
                font = ImageFont.load_default()
    except Exception:
        font = ImageFont.load_default()

    # --- Draw each character with slight random jitter -------------------
    char_spacing = width // (len(text) + 1)
    for i, ch in enumerate(text):
        x = char_spacing * i + random.randint(-4, 4) + 10
        y_center = (height - font_size) // 2
        y = y_center + (random.randint(-8, 8) if distort else 0)
        draw.text((x, y), ch, font=font, fill=text_color)

    # --- Noise layers ----------------------------------------------------
    add_noise_dots(draw, width, height, count=blue_dot_count, color=(0, 0, 255))
    add_noise_squares(draw, width, height, count=black_square_count, color=(0, 0, 0))

    # --- Very slight blur to blend noise with text (optional) ------------
    img = img.filter(ImageFilter.SMOOTH_MORE)

    img.save(output_path)
    print(f"CAPTCHA saved → {output_path}  |  text: '{text}'")
    return img, text


# ---------------------------------------------------------------------------
# Demo: generate a few sample CAPTCHAs
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    import os

    out_dir = "captcha_output"
    os.makedirs(out_dir, exist_ok=True)

    for i in range(5):
        img, text = generate_captcha(
            output_path=os.path.join(out_dir, f"captcha_{i+1}.png"),
            width=320,
            height=80,
            font_size=50,
            blue_dot_count=90,
            black_square_count=15,
            distort=True,
        )
        print(f"Displaying CAPTCHA with text: '{text}'")
        display(img)

    print(f"\n5 CAPTCHAs saved in ./{out_dir}/")
```
