CAPTCHA GENERATION

```python
import random
import string
from PIL import Image, ImageDraw, ImageFont, ImageFilter
from IPython.display import display # Import display for showing images in Colab
import math # Required for sinusoidal warping

def generate_captcha_text(length=8, chars=None):
    """Generate a random CAPTCHA string."""
    if chars is None:
        # Mix of digits, lowercase, uppercase — avoid visually ambiguous chars
        chars = string.digits + string.ascii_lowercase + string.ascii_uppercase
        chars = chars.replace('0', '').replace('O', '').replace('l', '').replace('I', '1')
    return ''.join(random.choices(chars, k=length))


def add_noise_dots(draw, width, height, count=80, color=(0, 0, 255)):
    """Add random scattered dots (like the blue dots in the sample)."""
    for _ in range(count):
        x = random.randint(0, width)
        y = random.randint(0, height)
        radius = random.randint(1, 3)
        draw.ellipse([x - radius, y - radius, x + radius, y + radius], fill=color)


def add_noise_squares(draw, width, height, count=10, color=(0, 0, 0)):
    """Add random small black squares as extra noise."""
    for _ in range(count):
        x = random.randint(0, width - 6)
        y = random.randint(0, height - 6)
        size = random.randint(3, 7)
        draw.rectangle([x, y, x + size, y + size], fill=color)


def add_curved_lines(draw, width, height, count=3, color=(100, 100, 100)):
    """Add random curved lines to the CAPTCHA for complexity."""
    for _ in range(count):
        # Random start and end points for the line
        x1, y1 = random.randint(0, width), random.randint(0, height)
        x2, y2 = random.randint(0, width), random.randint(0, height)

        # Add control points for curvature, slightly outside the image sometimes
        control_x1 = random.randint(-width // 4, width + width // 4)
        control_y1 = random.randint(-height // 4, height + height // 4)
        control_x2 = random.randint(-width // 4, width + width // 4)
        control_y2 = random.randint(-height // 4, height + height // 4)

        # Draw Bezier curve (quadratic or cubic)
        draw.line([(x1, y1), (control_x1, control_y1), (x2, y2)], fill=color, width=random.randint(1, 2))


def generate_captcha(
    text=None,
    width=300,
    height=80,
    bg_color=(255, 255, 255),
    text_color=(0, 0, 0),
    font_size=52,
    font_path=None,          # Path to a .ttf font file; None = PIL default
    blue_dot_count=80,
    black_square_count=12,
    curved_line_count=3,     # New parameter for curved lines
    randomize_bg_color=False, # New parameter to randomize background color
    distort=True,
    rotate_chars=False,      # Parameter for character rotation
    warp_chars=False,        # NEW parameter for character warping
    output_path="captcha.png",
):
    """
    Generate a CAPTCHA image.

    Args:
        text            : The CAPTCHA string (auto-generated if None).
        width / height  : Image dimensions in pixels.
        bg_color        : Background colour (RGB tuple).
        text_color      : Main text colour (RGB tuple).
        font_size       : Point size of the text.
        font_path       : Path to a TrueType/OpenType font file.
        blue_dot_count  : Number of blue noise dots.
        black_square_count: Number of black noise squares.
        curved_line_count : Number of curved lines to add.
        randomize_bg_color: If True, generates a random light background color.
        distort         : Apply slight per-character vertical jitter.
        rotate_chars    : If True, individual characters are randomly rotated.
        warp_chars      : If True, individual characters are warped with a sine wave.
        output_path     : Where to save the final PNG.

    Returns:
        (image, text) — PIL Image object and the CAPTCHA string used.
    """
    if text is None:
        text = generate_captcha_text()

    if randomize_bg_color:
        # Generate a light random background color for readability
        r = random.randint(200, 255)
        g = random.randint(200, 255)
        b = random.randint(200, 255)
        bg_color = (r, g, b)

    img = Image.new("RGB", (width, height), color=bg_color)
    draw = ImageDraw.Draw(img)

    # --- Load font -------------------------------------------------------
    try:
        if font_path:
            font = ImageFont.truetype(font_path, font_size)
        else:
            # Try common system mono fonts before falling back to PIL default
            for candidate in [
                "/usr/share/fonts/truetype/dejavu/DejaVuSansMono-Bold.ttf",
                "/usr/share/fonts/truetype/liberation/LiberationMono-Bold.ttf",
                "/Library/Fonts/Courier New Bold.ttf",
                "C:/Windows/Fonts/courbd.ttf",
            ]:
                try:
                    font = ImageFont.truetype(candidate, font_size)
                    break
                except IOError:
                    continue
            else:
                font = ImageFont.load_default()
    except Exception:
        font = ImageFont.load_default()

    # --- Draw each character with slight random jitter and optional rotation/warping ---
    char_spacing = width // (len(text) + 1)
    for i, ch in enumerate(text):
        x_base = char_spacing * i + 10
        y_center = (height - font_size) // 2
        y_jitter = (random.randint(-8, 8) if distort else 0)

        # Get character bounding box
        try:
            bbox = font.getbbox(ch)
        except AttributeError:
            # Fallback for older Pillow versions or if bbox is not directly available
            mask = font.getmask(ch)
            bbox = mask.getbbox() if mask else (0, 0, 0, 0)

        char_w = bbox[2] - bbox[0]
        char_h = bbox[3] - bbox[1]

        # Create a temporary image for the character, with padding for rotation/warping
        temp_img_size_factor = 0.5
        # Ensure enough space for character + padding for rotation/warp
        temp_img_width = max(char_w, char_h) + int(font_size * temp_img_size_factor)
        temp_img_height = max(char_w, char_h) + int(font_size * temp_img_size_factor)

        # Handle cases where char_w or char_h might be zero or very small
        if temp_img_width <= 0: temp_img_width = font_size + int(font_size * temp_img_size_factor)
        if temp_img_height <= 0: temp_img_height = font_size + int(font_size * temp_img_size_factor)

        temp_char_img = Image.new("RGBA", (temp_img_width, temp_img_height), (0, 0, 0, 0))
        temp_char_draw = ImageDraw.Draw(temp_char_img)

        # Calculate position to draw char on temp_char_img, centering it
        draw_x = (temp_img_width - char_w) // 2 - bbox[0]
        draw_y = (temp_img_height - char_h) // 2 - bbox[1]
        temp_char_draw.text((draw_x, draw_y), ch, font=font, fill=text_color)

        final_char_img = temp_char_img

        if rotate_chars:
            angle = random.randint(-30, 30) # Random rotation angle
            final_char_img = final_char_img.rotate(angle, expand=1, resample=Image.BICUBIC)

        if warp_chars:
            # Apply a simple sinusoidal warp
            warp_amplitude = random.randint(2, 6) # Pixels of horizontal displacement
            warp_frequency = random.uniform(0.05, 0.2) # How many waves per character height

            warped_char_img = Image.new("RGBA", final_char_img.size, (0, 0, 0, 0))
            pixels = final_char_img.load()
            warped_pixels = warped_char_img.load()

            for y_coord in range(final_char_img.height):
                # Calculate horizontal displacement based on a sine wave
                x_offset = int(warp_amplitude * math.sin(warp_frequency * y_coord))
                for x_coord in range(final_char_img.width):
                    # Sample pixel from original image with offset
                    original_x = x_coord - x_offset
                    if 0 <= original_x < final_char_img.width:
                        warped_pixels[x_coord, y_coord] = pixels[original_x, y_coord]
            final_char_img = warped_char_img

        # Calculate original character's target center on the main image
        target_center_x = x_base + random.randint(-4, 4) + char_w / 2
        target_center_y = y_center + y_jitter + char_h / 2

        # Calculate where to paste the final character's top-left
        p_w, p_h = final_char_img.size
        paste_x = int(target_center_x - p_w / 2)
        paste_y = int(target_center_y - p_h / 2)

        img.paste(final_char_img, (paste_x, paste_y), final_char_img) # Use final_char_img as mask for alpha

    # --- Noise layers ----------------------------------------------------
    add_noise_dots(draw, width, height, count=blue_dot_count, color=(0, 0, 255))
    add_noise_squares(draw, width, height, count=black_square_count, color=(0, 0, 0))
    add_curved_lines(draw, width, height, count=curved_line_count, color=(100, 100, 100))

    # --- Very slight blur to blend noise with text (optional) ------------
    img = img.filter(ImageFilter.SMOOTH_MORE)

    img.save(output_path)
    print(f"CAPTCHA saved → {output_path}  |  text: '{text}'")
    return img, text


# ---------------------------------------------------------------------------
# Demo: generate a few sample CAPTCHAs
# ---------------------------------------------------------------------------
if __name__ == "__main__":
    import os

    out_dir = "captcha_output"
    os.makedirs(out_dir, exist_ok=True)

    print("\n--- Generating CAPTCHAs with Curved Lines, Random Backgrounds, Character Rotation, and Warping---")
    for i in range(5):
        img, text = generate_captcha(
            output_path=os.path.join(out_dir, f"captcha_{i+1}.png"),
            width=320,
            height=80,
            font_size=50,
            blue_dot_count=90,
            black_square_count=15,
            curved_line_count=3,
            randomize_bg_color=True,
            distort=True,
            rotate_chars=True, # Enable character rotation
            warp_chars=True,   # Enable character warping
        )
        print(f"Displaying CAPTCHA with text: '{text}'")
        display(img)

    print(f"\n5 CAPTCHAs saved in ./{out_dir}/\n")
```

### Download CAPTCHA Images as a Zip File

```python
import zipfile
from google.colab import files

zip_file_name = 'captcha_images.zip'

# Create a ZipFile object
with zipfile.ZipFile(zip_file_name, 'w') as zipf:
    for root, dirs, files_in_dir in os.walk(out_dir):
        for file in files_in_dir:
            file_path = os.path.join(root, file)
            zipf.write(file_path, arcname=os.path.relpath(file_path, out_dir))

print(f"'{zip_file_name}' created successfully. You can download it using the link below:")
files.download(zip_file_name)
```
